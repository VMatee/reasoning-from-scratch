# Exact arithmetic search + verifiable traces

This demonstration explores candidate arithmetic reasoning steps using `fractions.Fraction` and independently verifies each step. It is **not a trained language model**.

```python
from fractions import Fraction
from verifier_reasoning import solve, verify_steps, Step

numbers = [3, 3, 8, 8]
result = solve(numbers, target=24)
print(result.expression)
for step in result.steps:
    print(step)
print("Verified:", verify_steps(numbers, result.steps))
```

## Why a verifier matters

A plausible-looking trace can still contain incorrect arithmetic. We alter the final answer and observe the independent checker reject it.

```python
fake = list(result.steps)
fake[-1] = Step(fake[-1].left, fake[-1].operator, fake[-1].right, Fraction(25))
assert verify_steps(numbers, fake, target=24) is False
print("Tampered trace correctly rejected")
```

## Experiment

Change the input numbers, target and search-state budget. A missing answer with a small budget does not prove there is no solution.

```python
for values in ([1, 2, 3, 4], [2, 3, 4, 5]):
    found = solve(values, target=24, max_states=20_000)
    print(values, found.expression if found else "no verified solution within budget")
```
